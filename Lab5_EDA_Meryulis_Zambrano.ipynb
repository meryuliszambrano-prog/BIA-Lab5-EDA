````python
#@title  Geodataframe For Maps
import os
import geopandas as gpd

# The dataset only has the .shp file (no .shx / .dbf),
# so we ask GDAL to rebuild the .shx index automatically
os.environ['SHAPE_RESTORE_SHX'] = 'YES'
````

````python
map_df = gpd.read_file("London_Borough_Excluding_MHW.shp")
# check data type so we can see that this is not a normal dataframe, but a GEOdataframe
print(type(map_df))
map_df.head()
````

````python
# Map of the London boroughs
map_df.plot(figsize=(10,10), color='lightblue', edgecolor='black')
plt.title('London Boroughs')
plt.show()
````

````python
map_df
````

````python
# Without the .dbf file there are no names, so we calculate the area of each borough
# (coordinates are in metres -> area in km2) and draw a choropleth map
map_df['area_km2'] = map_df.geometry.area / 1e6
map_df.plot(column='area_km2', cmap='Blues', legend=True, edgecolor='black', figsize=(10,10))
plt.title('London Boroughs by area (km2)')
plt.show()
````
